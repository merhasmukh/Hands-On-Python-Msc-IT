```python
import json
import pandas as pd

# Step 1: Accept the number of students
n = int(input("Enter the number of students: "))

students = []

# Step 2: Accept student details
for i in range(n):

    print(f"\nEnter details for Student {i + 1}")

    name = input("Student Name: ")
    age = int(input("Age: "))
    course = input("Course: ")

    student = {
        "Name": name,
        "Age": age,
        "Course": course
    }

    students.append(student)

# Step 3: Store student records in a JSON file
with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

print("\nStudent records saved successfully to students.json")


# Step 4: Read student records from the JSON file
with open("students.json", "r") as file:
    data = json.load(file)

print("\n--- All Student Information ---")

for student in data:
    print(f"Name   : {student['Name']}")
    print(f"Age    : {student['Age']}")
    print(f"Course : {student['Course']}")
    print()


# Step 5: Load JSON data into a Pandas DataFrame
df = pd.DataFrame(data)

print("--- Pandas DataFrame ---")
print(df)


# Step 6: Display first two records
print("\n--- First Two Student Records ---")
print(df.head())


# Step 7: Display last two records
print("\n--- Last Two Student Records ---")
print(df.tail())


# Step 8: Display shape
print("\n--- Shape of DataFrame ---")
print(df.shape)


# Step 9: Display column names
print("\n--- Column Names ---")
print(df.columns.tolist())


# Step 10: Display data types
print("\n--- Data Types ---")
print(df.dtypes)
```
